```python
from transformers import AutoTokenizer, AutoModelForTokenClassification
from transformers import pipeline
```

```python
tokenizer = AutoTokenizer.from_pretrained("dslim/bert-base-NER")
model = AutoModelForTokenClassification.from_pretrained("dslim/bert-base-NER")
```

```python
nlp = pipeline("ner", model=model, tokenizer=tokenizer)
```

```python
def NER(text):
  return nlp(text)
```

```python
NER("My name is Wolfgang and I live in Berlin. my brother name is india and he lives in chennai.")
```

```python
NER("The man name is koo and he is flying in the sky.")
```

```python
NER("Google dont like iphone")
```

```python
NER("Apple Inc. was founded by Steve Jobs in California.")
```

```python
NER("The Iphone is big.")
```
